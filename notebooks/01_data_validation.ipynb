## Review the raw validation output

This summary should confirm the raw file matches the expected shape and catches the known structural issue: blank values in TotalCharges for customers with tenure equal to 0.

```python
blank_total_charges = raw_df['TotalCharges'].astype(str).str.strip().eq('')
tenure_zero = raw_df['tenure'].astype(str).str.strip().eq('0')
print('blank TotalCharges rows:', int(blank_total_charges.sum()))
print('tenure = 0 rows:', int(tenure_zero.sum()))
raw_df.loc[blank_total_charges, ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']]
```
